```python
import os
import pandas as pd
import numpy as np

# Create directory if it doesn't exist
os.makedirs('/content/extracted_data', exist_ok=True)

# Generate synthetic student performance data
np.random.seed(42)
n_students = 100

final_exam_scores = np.random.normal(75, 12, n_students).clip(50, 100).astype(int)

def assign_grade(score):
    if score >= 90: return 'A'
    elif score >= 80: return 'B'
    elif score >= 70: return 'C'
    elif score >= 60: return 'D'
    else: return 'F'

final_grades = [assign_grade(s) for s in final_exam_scores]

df_synthetic = pd.DataFrame({
    'student_id': range(1, n_students + 1),
    'final_exam_score': final_exam_scores,
    'final_grade': final_grades
})

# Save the generated dataset
df_synthetic.to_csv('/content/extracted_data/student_performance_dataset.csv', index=False)
print("Synthetic dataset successfully created at '/content/extracted_data/student_performance_dataset.csv'!")
```

DIFFERENT PLOTTING TECHNIQUES

```python
import pandas as pd

df = pd.read_csv('/content/extracted_data/student_performance_dataset.csv')

display(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['final_exam_score'], bins=15, kde=True)
plt.title('Distribution of Final Exam Scores')
plt.xlabel('Final Exam Score')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
display(df.head())
```

```python
final_grade_distribution = df['final_grade'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(final_grade_distribution, labels=final_grade_distribution.index, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
plt.title('Distribution of Student Final Grades')
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['final_exam_score'])
plt.title('Box Plot of Final Exam Scores')
plt.ylabel('Final Exam Score')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
