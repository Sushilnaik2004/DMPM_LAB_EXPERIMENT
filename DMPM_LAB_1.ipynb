```python
# Mean

data = [10, 20, 30, 40, 50]

m = 0
for i in data:
    m += i

mean = m / len(data)

print("Mean is:", mean)
```

```python
# Median

data = [10, 20, 40, 60, 80]

data.sort()

n = len(data)

if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]

print("Median is:", median)
```

```python
# Mode

nums = [1, 2, 2, 3, 3, 3, 4, 5, 5]

counts = {}

for i in nums:
    if i in counts:
        counts[i] += 1
    else:
        counts[i] = 1

max_count = 0

for num, count in counts.items():
    if count > max_count:
        max_count = count
        mode = num

print("The Mode is:", mode)
```

```python
# Variance

nums = [10, 20, 30, 40, 50]

mean = sum(nums) / len(nums)

x = 0

for i in nums:
    x += (i - mean) ** 2

variance = x / len(nums)

print("Variance is:", variance)
```

```python
# Standard Deviation

import math

nums = [10, 20, 30, 40, 50]

mean = sum(nums) / len(nums)

x = 0

for i in nums:
    x += (i - mean) ** 2

variance = x / len(nums)

std = math.sqrt(variance)

print("Standard Deviation is:", std)
```

```python
# with function
def mean(nums):
    sum = 0

    for i in nums:
        sum += i

    return sum / len(nums)


nums = [10, 20, 30, 40, 50]

result = mean(nums)

print("Mean =", result)
```

```python
def median(nums):
    nums.sort()
    n = len(nums)

    if n % 2 == 0:
        return (nums[n//2 - 1] + nums[n//2]) / 2
    else:
        return nums[n//2]


nums = [10, 20, 30, 40, 50]

result = median(nums)

print("Median =", result)
```

```python
def mode(nums):
    freq = {}

    for i in nums:
        freq[i] = freq.get(i, 0) + 1

    max_count = 0
    mode_value = None

    for key, value in freq.items():
        if value > max_count:
            max_count = value
            mode_value = key

    return mode_value


nums = [10, 20, 30, 40, 50]

result = mode(nums)

print("Mode =", result)
```

```python
def variance(nums):
    total = 0
    mean = sum(nums) / len(nums)

    for i in nums:
        total += (i - mean) ** 2

    return total / len(nums)


nums = [10, 20, 30, 40, 50]

result = variance(nums)

print("Variance =", result)
```
