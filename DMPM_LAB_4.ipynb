NORMALIZATION

```python
#min-max
data = [10, 20, 30, 40, 50,]

min_val = min(data)
max_val = max(data)

normalized = [(x - min_val) / (max_val - min_val) for x in data]

print(normalized)
```

```python
#z-score
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)
std = (sum((x - mean)**2 for x in data) / len(data)) ** 0.5

z_scores = [(x - mean) / std for x in data]

print(z_scores)
```

```python
#Decimal scaling
data = [100, 200, 300, 400]

j = len(str(max(data)))
normalized = [x / (10 ** j) for x in data]

print(normalized)
```

```python
import numpy as np

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

covariance = np.cov(x, y)[0][1]
correlation = np.corrcoef(x, y)[0][1]

print("Covariance:", covariance)
print("Correlation:", correlation)
```
